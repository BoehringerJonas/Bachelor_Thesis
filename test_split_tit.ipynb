```python
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
```

```python
signal = np.load(r"D:\BA\workspace\data\clean\train\CPAP-005-4\TITeit.npz")
```

```python
sig = signal["pressureData"][0,:]
ges = len(signal["pressureData"][0])
schwelle = 2
länge = 400
step_point = np.zeros(ges)
step_point.shape
```

```python
sig = savgol_filter(sig, 52, 3)
```

```python
for i in range(ges - 2*länge):
    p1 = np.min(sig[i:i+länge])# / länge
    p2 = np.min(sig[i+länge:i+2*länge])# / länge
    if p1-p2 >= schwelle:
        step_point[i + länge] = sig[i + länge]

step_point.shape
```

```python
time = np.arange(0, len(step_point))
time_ges = np.arange(0, ges)

plt.plot(time_ges, sig)
plt.plot(time, step_point, color="red")

plt.show()
```

```python
count = 0
for j in range(ges):
    if step_point[j] > 0:
        count += 1
        print(f"{step_point[j]} + {j}")

print(count)
```
